### Loading and Preprocessing Student Performance Data

```python
import pandas as pd

# Load the newly uploaded dataset
student_df = pd.read_csv('/content/Student_Performance.csv')

x=student_df.iloc[:,:-1]
y=student_df.iloc[:,-1]
print(x)
print(y)
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize and fit the LabelEncoder on 'Extracurricular Activities'
le = LabelEncoder()

x['Extracurricular Activities'] = le.fit_transform(x['Extracurricular Activities'])
print(x)
```

### 1. Train-Test Split for Student Performance Data
We will split the feature matrix `x` and the target vector `y` into training (80%) and testing (20%) sets.

```python
from sklearn.model_selection import train_test_split

# Split the dataset into 80% training and 20% testing
X_train_student, X_test_student, y_train_student, y_test_student = train_test_split(
    x, y, test_size=0.2, random_state=42
)

print("Training set shape:", X_train_student.shape)
print("Testing set shape:", X_test_student.shape)
```

### 2. Model Training and Evaluation
We will train a simple **Linear Regression** model and a **Decision Tree Regressor** to predict the Performance Index.

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_absolute_error

# 1. Linear Regression
lr_student = LinearRegression()
lr_student.fit(X_train_student, y_train_student)
y_pred_lr = lr_student.predict(X_test_student)

# 2. Decision Tree Regressor
dt_student = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_student.fit(X_train_student, y_train_student)
y_pred_dt = dt_student.predict(X_test_student)

# Evaluation Metrics
print("--- Linear Regression Performance ---")
print(f"R2 Score: {r2_score(y_test_student, y_pred_lr):.4f}")
print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test_student, y_pred_lr):.2f}")

print("\n--- Decision Tree Performance ---")
print(f"R2 Score: {r2_score(y_test_student, y_pred_dt):.4f}")
print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test_student, y_pred_dt):.2f}")
```

### 3. Random Forest Regression on Student Performance Data
Let's train a Random Forest Regressor to see how an ensemble method performs on this dataset.

```python
from sklearn.ensemble import RandomForestRegressor

# Initialize and train the Random Forest Regressor
rf_student = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
rf_student.fit(X_train_student, y_train_student)

# Predict on the test set
y_pred_rf_student = rf_student.predict(X_test_student)

# Calculate and display metrics
print("--- Random Forest Performance ---")
print(f"R2 Score: {r2_score(y_test_student, y_pred_rf_student):.4f}")
print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test_student, y_pred_rf_student):.2f}")
```

### Loading and Preprocessing Insurance Dataset

```python
import pandas as pd

# Load the insurance dataset
insurance_df = pd.read_csv('/content/insurance.csv')

# Let's inspect the first few rows to see the column structure
display(insurance_df.head())
print(insurance_df.info())
```

### Encoding Categorical Variables
We need to convert non-numeric columns like `sex`, `smoker`, and `region` to numeric values before training our regression models.

```python
from sklearn.preprocessing import OneHotEncoder

# Make a copy for features and target
X_ins = insurance_df.drop(columns=['charges'])
y_ins = insurance_df['charges']

# Get categorical columns
categorical_cols = X_ins.select_dtypes(include=['object']).columns.tolist()

# Simple One-Hot Encoding using sklearn
ohe = OneHotEncoder(drop='first', sparse_output=False)
ohe_df = pd.DataFrame(ohe.fit_transform(X_ins[categorical_cols]), columns=ohe.get_feature_names_out(categorical_cols))

# Combine back with the numerical features
X_ins = X_ins.drop(columns=categorical_cols).reset_index(drop=True)
X_ins = pd.concat([X_ins, ohe_df], axis=1)

display(X_ins.head())
```

### Train-Test Split for Insurance Data

```python
from sklearn.model_selection import train_test_split

# Split into 80% training and 20% testing
X_train_ins, X_test_ins, y_train_ins, y_test_ins = train_test_split(
    X_ins, y_ins, test_size=0.2, random_state=42
)

print("Training set shape:", X_train_ins.shape)
print("Testing set shape:", X_test_ins.shape)
```

### Training and Evaluating Regression Models

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error

# 1. Linear Regression
lr_ins = LinearRegression()
lr_ins.fit(X_train_ins, y_train_ins)
y_pred_lr_ins = lr_ins.predict(X_test_ins)

# 2. Decision Tree Regressor
dt_ins = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_ins.fit(X_train_ins, y_train_ins)
y_pred_dt_ins = dt_ins.predict(X_test_ins)

# 3. Random Forest Regressor
rf_ins = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
rf_ins.fit(X_train_ins, y_train_ins)
y_pred_rf_ins = rf_ins.predict(X_test_ins)

# Print Evaluation Metrics
print("--- Linear Regression ---")
print(f"R2 Score: {r2_score(y_test_ins, y_pred_lr_ins):.4f}")
print(f"MAE: {mean_absolute_error(y_test_ins, y_pred_lr_ins):,.2f}")

print("\n--- Decision Tree (depth=5) ---")
print(f"R2 Score: {r2_score(y_test_ins, y_pred_dt_ins):.4f}")
print(f"MAE: {mean_absolute_error(y_test_ins, y_pred_dt_ins):,.2f}")

print("\n--- Random Forest (depth=5) ---")
print(f"R2 Score: {r2_score(y_test_ins, y_pred_rf_ins):.4f}")
print(f"MAE: {mean_absolute_error(y_test_ins, y_pred_rf_ins):,.2f}")
```
